```python

from sklearn.tree import DecisionTreeClassifier

# Sample training data
X = [[1], [2], [3], [4]]
y = [0, 0, 1, 1]

# Create and train the model
model = DecisionTreeClassifier(random_state=0)
model.fit(X, y)

# Predict a new value
prediction = model.predict([[3]])

print("Machine Learning Example")
print("Input value:", 3)
print("Predicted class:", prediction[0])
```

```python
## Relationship Between AI, ML and Deep Learning

Artificial Intelligence is the broad field of creating intelligent computer systems. Machine Learning is a subset of AI that allows computers to learn from data. Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers.

Therefore:

- Deep Learning is part of Machine Learning.
- Machine Learning is part of Artificial Intelligence.

Example: A facial recognition system uses AI techniques, may use ML to learn from images, and can use Deep Learning to identify complex facial patterns.
```
